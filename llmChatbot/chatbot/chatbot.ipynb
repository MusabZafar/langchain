## Message History
We can use a Message History class to wrap our model and make it stateful. 
This will keep track of inputs and outputs of the model, and store them in 
some datastore. Future interactions will then load those messages and pass 
them into the chain as part of the input. Let's see how to use this!

```python
!pip install langchain_community
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}
def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id] 
```

```python
with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable": {"session_id": "musab_chat_1"}}
```

```python
response1 = with_message_history.invoke(
    [HumanMessage(content="Hi, I am Musab Zafar and I am learning LangChain with Groq.")],
    config=config
)
print(response1.content)
```

```python
response2 = with_message_history.invoke(
    [HumanMessage(content="What is my name and what am I learning?")],
    config=config
)
print(response2.content)
```

## Managing the conversation history

```python
from operator import itemgetter
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, trim_messages
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables import RunnablePassthrough
from langchain_groq import ChatGroq


model = ChatGroq(
    model="openai/gpt-oss-120b",
    max_tokens=500
)


trimer = trim_messages(
    max_tokens=70,
    strategy="last",
    token_counter=model.get_num_tokens_from_messages,
    include_system=True,
    allow_partial=False,
    start_on="human"
)


messages = [
    SystemMessage(content="You are a helpful AI assistant."),
    HumanMessage(content="Hi! My name is Musab and I am building a chatbot in Python."),
    AIMessage(content="Hello Musab! That sounds great. Are you using LangChain and Groq?"),
    HumanMessage(content="Yes, I am using LangChain with Groq and learning how memory trimming works."),
    AIMessage(content="Trimming keeps your context window small by dropping older messages."),
    HumanMessage(content="What is my name and what am I building?")
]


prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. You must reply strictly in native {language}."),
    MessagesPlaceholder(variable_name="messages"),
])


chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages") | trimer)
    | prompt
    | model
)


response = chain.invoke(
    {
        "messages": messages + [HumanMessage(content="What is my name?")],
        "language": "English"
    }
)

print(response.content)
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

# 1. In-memory store dictionary to hold session histories
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
        # Optional: Seed the new session with your system message if needed
        store[session_id].add_message(SystemMessage(content="You are a helpful AI assistant."))
    return store[session_id]

# 2. Wrap your existing chain with message history management
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages",
    history_messages_key="messages"
)

# 3. Configure a unique session ID
config = {"configurable": {"session_id": "musab_trimmed_session_1"}}

# 4. First turn: Introduce yourself
response1 = with_message_history.invoke(
    {
        "messages": [HumanMessage(content="Hi! My name is Musab and I am building a chatbot in Python.")],
        "language": "English"
    },
    config=config
)
print("Assistant:", response1.content)

# 5. Second turn: Ask a follow-up to test if memory recall and trimming work smoothly
response2 = with_message_history.invoke(
    {
        "messages": [HumanMessage(content="What is my name and what am I building?")],
        "language": "English"
    },
    config=config
)
print("Assistant:", response2.content)
```
