### Messages
Messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:

• Role - Identifies the message type (e.g. system, user)

• Content - Represents the actual content of the message (like text, images, audio, documents, etc.)

• Metadata - Optional fields such as response information, message IDs, and token usage

LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:openai/gpt-oss-20b")
```

```python
model.invoke("Please tell me what is AI")
```

### Text Prompts
In LangChain, text prompts are plain strings passed directly to a chat model when you only need a single response without managing multi-turn conversation history or complex message roles.

```python
model.invoke("What is langchain")
```

### Message Prompts
In LangChain, message prompts are structured lists of message objects passed to a chat model to define system instructions, manage multi-turn conversation history, or send multimodal inputs.

• Conversation Memory. Chat models are stateless by default, so passing the full list of prior HumanMessage and AIMessage turns gives the model context for follow-up questions.

• Role Separation. Separating system instructions from user queries helps prevent prompt injection and keeps the model aligned with your formatting rules.


### System Message
In LangChain, a System Message (SystemMessage) sets the overall behavior, persona, and operational rules for a chat model before the user interaction begins. It acts as persistent background instructions that guide how the model responds throughout the conversation.

• Place it first. Always put SystemMessage at the very beginning (index 0) of your messages list so the model processes your rules before reading user input.

• Keep instructions clear and direct. Specific constraints (like "Respond in under 50 words") work much better than vague requests (like "Be helpful").



### AI Message 

An AIMessage represents any output generated by the chat model. It is returned automatically whenever you call model.invoke().

• Text Responses. When the model answers directly, the generated text is stored in response.content (or response.text).

• Tool Call Requests. When the model decides to use a tool, response.tool_calls contains a list of structured requests specifying the tool name, arguments, and a unique id.

• Metadata. It carries response details such as usage_metadata (input/output token counts) and response_metadata (model name, finish reason).

• Conversation Memory. You can also instantiate an AIMessage("...") manually to inject few-shot examples or restore previous assistant replies into your messages list.

### Tool Message

A ToolMessage carries the output of a Python function back to the model after your application executes a requested tool call.

• Links to the AI Request. Every ToolMessage requires a tool_call_id that matches the id from the model's AIMessage.tool_calls. This tells the model which specific tool request produced which result.

• Automatic Creation. When you pass a tool_call object directly into a LangChain @tool function using my_tool.invoke(tool_call), LangChain automatically wraps the return value into a properly formatted ToolMessage.


```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on artificial intelligence")
]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]
response = model.invoke(messages)
print(response.content)
```

```python
## Detailed info to the LLM through System message
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]
response = model.invoke(messages)
print(response.content)
```

```python
## Message Metadata
human_msg = HumanMessage(
    content="Hello!",
    name="musab",  # Optional: identify different users
    id="msg_123",  # Optional: unique identifier for tracing
)
```

```python
response=model.invoke([human_msg])
response
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# Create an AI message manually (e.g., for conversation history)
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Add to conversation history
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,  # Insert as if it came from the model
    HumanMessage("Great! What's 2+2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,   # Model's tool call
    tool_message, # Tool execution result
]
response = model.invoke(messages)  
```
